## 1. Operadores Aritméticos no Python

| Operador | Operação | Exemplo | Resultado |
| :---: | :--- | :--- | :---: |
| `+` | Adição | `10 + 5` | `15` |
| `-` | Subtração | `10 - 5` | `5` |
| `*` | Multiplicação | `10 * 5` | `50` |
| `/` | Divisão Real | `10 / 4` | `2.5` |
| `//` | Divisão Inteira | `10 // 4` | `2` |
| `%` | Módulo (Resto da Divisão) | `10 % 3` | `1` |
| `**` | Exponenciação (Potência) | `2 ** 3` | `8` |

> **Ordem de Precedência:** Parênteses `()` ➔ Exponenciação `**` ➔ Multiplicação/Divisão `*`, `/`, `//`, `%` ➔ Adição/Subtração `+`, `-`.


````python
# Exemplo de operações aritméticas
preco_original = 100.0
desconto = 15.0
preco_final = preco_original - desconto
parcelas = 3
valor_parcela = preco_final / parcelas

print("Preço final:", preco_final)
print("Valor de cada parcela (3x):", valor_parcela)
````

## 2. Entrada de Dados (`input`) e *Casting*
A função `input()` sempre retorna uma **string (`str`)**, mesmo que o usuário digite apenas números. Para realizar cálculos, devemos converter (*casting*) explicitamente:
* `int(dado)`: Converte para número inteiro.
* `float(dado)`: Converte para número decimal.
* `str(dado)`: Converte para texto.

### Saída Formatada com f-strings:
```python
nome = "Carlos"
nota = 8.75
print(f"O aluno {nome} obteve média {nota:.1f}")
```


````python
# Exemplo do Professor: Leitura e conversão
nome_vendedor = input("Digite o nome do vendedor: ")
vendas_str = input("Digite o total de vendas (R$): ")
total_vendas = float(vendas_str)

comissao = total_vendas * 0.05
print(f"Parabéns {nome_vendedor}! Sua comissão é de R$ {comissao:.2f}")
````
